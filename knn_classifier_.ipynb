```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import make_classification

X,y= make_classification(
    n_samples=1000,
    n_features=5,
    n_classes=2,
    n_redundant=0,
    n_clusters_per_class=1,
    random_state=999
)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=999)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

# here during the hyper parameter tuning : we have to select the value of the p ( where the value of the p = 1, when manhatten distance ; for the = 2 , euclidean distance )


similar , we also have to select algorithm during the hyper parameter tuning , for we have 3 algo : ball_tree, kd_tree, brute ,  another is 'auto' jisme 3 me se koi algo select hogi based upon the dataset

```python
classifier = KNeighborsClassifier(n_neighbors=5,algorithm='auto')
classifier.fit(X_train,y_train)
```

```python
y_pred =classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
a=0

for k in range(1, 11):
  classifier = KNeighborsClassifier(n_neighbors=k,algorithm='auto')
  classifier.fit(X_train,y_train)
  y_pred =classifier.predict(X_test)
  if a<accuracy_score(y_test,y_pred):
    a=accuracy_score(y_test,y_pred)
    print(a)
    print(k)
```
